```python
def calculate_grade(marks):
    if marks>=90:
        return "A"
    elif marks>=80:
        return "B"
    elif marks>=70:
        return "C"
    elif marks>=40:
        return "D"
    return "fail"
print(calculate_grade(96))
print(calculate_grade(65))
print(calculate_grade(30))

```
